```python
import random

def guess():
    answer:int = random.randint(0,1000)
    
    while True:
        try:
            guess = int(input('guess the number!'))
        except ValueError:
            print('chert nazan')
            continue
        if guess == answer:
            print('dayus')
            print(f'number was {answer}')
            break
        elif guess > answer:
            print('kochiktare')
        else:
            print('bozorgtare')
        
guess()
```
