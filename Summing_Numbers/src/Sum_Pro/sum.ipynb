```python
def sum_advanced() -> None:
    total: float = 0.0
    count: int = 0
    largest: float | None = None
    smallest: float | None = None

    while True:
        user: str = input('Enter a number: ').strip()
        if not user:
            break

        try:
            number: float = float(user)
        except ValueError:
            print('Invalid input')
            continue

        total += number
        count += 1

        if largest is None or number > largest:
            largest = number
        if smallest is None or number < smallest:
            smallest = number

    if count == 0:
        print('No numbers entered')
        return

    average = total / count

    print('-' * 30)
    print('Total    :', total)
    print('Count    :', count)
    print('Average  :', f'{average:.2f}')
    print('Largest  :', largest)
    print('Smallest :', smallest)
    print('-' * 30)
    
if __name__ == "__main__":
    sum_advanced()

```
