```python
import random

class Sang_kaghaz_gheychi:
    """baziah s

    :return: _description_
    :rtype: str
    """
    skg = ['sang','kaghaz', 'gheychi']
    
    
    def __init__(self,name):
        self.bazikon = name
        
    def eb(self):
        sb = input(f'begoo binam({Sang_kaghaz_gheychi.skg})')
        print(sb.lower())
        if sb.lower() in Sang_kaghaz_gheychi.skg:
            return sb.lower()
        
        print(f'dorost begoo palasht {Sang_kaghaz_gheychi.skg}.')
        return self.eb()
    def ep(self):
        return random.choice(Sang_kaghaz_gheychi.skg)
    
    def ew(self,eb,ep):
        if eb == ep:
            return ' Barabar'
        
        TB = [('sang','gheychi'),('kaghaz','sang'),('gheychi','kaghaz')]
        for TB in TB:
            if (eb == TB[0]) and (ep ==TB[1]):
                return ' Bordi Gozo'
        return ' Bega Rafti '
    
    def play(self):
        
        user = self.eb()
        computer = self.ep()
        print(f'ep:{computer}')
        print(self.ew(user,computer))    
        
game = Sang_kaghaz_gheychi('Emad')    
game.play()
```
